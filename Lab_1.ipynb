### Problem1. Write a program that inputs two numbers and displays their sum, difference, product and quotient.Convert the inputs to numeric values before calculating.

```python
# input two numbers with the basic arithmatic operation
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

sum_val = num1 + num2
difference = num1 - num2
product = num1 * num2
quotient = num1 / num2 if num2 != 0 else "Undefined (cannot divide by zero)"

print("\n--- Arithmetic Results ---")
print("Sum:", sum_val)
print("Difference:", difference)
print("Product:", product)
print("Quotient:", quotient)
```

### Problem 2. Input a student's name, student ID, age department and CGPA. Display the information as a formatted student profile.

```python

name = input("Enter Student Name: ")
student_id = input("Enter Student ID: ")
age = int(input("Enter Age: "))
department = input("Enter Department: ")
cgpa = float(input("Enter CGPA: "))


print("\n==============================")
print("       STUDENT PROFILE        ")
print("==============================")
print(f"Name       : {name}")
print(f"Student ID : {student_id}")
print(f"Age        : {age}")
print(f"Department : {department}")
print(f"CGPA       : {cgpa:.2f}")
print("==============================")
```

### Problem 3. Input a temperature in Celsius and convert it to Fahrenheit and Kelvin using the formulas:

###                   F=(C×9/5)+32,   K=C+273.1

```python

celsius = float(input("Enter temperature in Celsius: "))

fahrenheit = (celsius * 9 / 5) + 32
kelvin = celsius + 273.15

print("\n--- Temperature Conversion ---")
print(f"Celsius    : {celsius:.2f} °C")
print(f"Fahrenheit : {fahrenheit:.2f} °F")
print(f"Kelvin     : {kelvin:.2f} K")
```

### Problem 4. Input a student’s name and five course marks.Calculate the total, average, highest mark, lowest mark and number of passed courses. Display all results. A mark of 50 or higher is considered a pass.

```python

student_name = input("Enter student name: ")

marks = []
for i in range(1, 6):
    m = float(input(f"Enter mark for Course {i}: "))
    marks.append(m)
    
total_marks = sum(marks)
average_mark = total_marks / len(marks)
highest_mark = max(marks)
lowest_mark = min(marks)

passed_count = sum(1 for m in marks if m >= 50)


print("\n========================================")
print(f"     PERFORMANCE REPORT: {student_name.upper()}     ")
print("========================================")
print("Marks Entered   :", marks)
print(f"Total Marks     : {total_marks:.2f}")
print(f"Average Mark    : {average_mark:.2f}")
print(f"Highest Mark    : {highest_mark:.2f}")
print(f"Lowest Mark     : {lowest_mark:.2f}")
print(f"Passed Courses  : {passed_count} / {len(marks)}")
print("========================================")
```
