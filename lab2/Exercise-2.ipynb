### Problem_1  –  Create a list containing the marks of 5 courses. Display the total marks, average mark, highest mark, and lowest mark. Also count how many courses the student passed (pass mark = 50).

```python
## Exercise 1: Weekend Temperature with a Tuple

programming_ds = float(input("enter your programming for ds mark : "))
data_visualization = float(input("enter your data visualization mark : "))
data_processing = float(input("enter your data processing mark : "))
data_science = float(input("enter your data science mark : "))
java = float(input("enter your java mark : "))

marks = [programming_ds, data_visualization, data_processing, data_science, java]
print("marks are : ", marks)

passed = 0
failed = 0
total = 0.0
maximum = marks[0]
minimum = marks[0]

for mark in marks:
    total += mark
    if mark > maximum:
        maximum = mark
    if mark < minimum:
        minimum = mark
    if mark >= 50:         
        passed += 1
    else:
        failed += 1

average = total / len(marks)

print("total marks are : ", total)
print("average mark is : ", average)
print("maximum mark is : ", maximum)
print("minimum mark is : ", minimum)
print("number of courses passed : ", passed)
print("number of courses failed : ", failed)
```

### Problem_2  –  Create a tuple containing the temperatures recorded on Saturday and Sunday for three different times of the day. Use a for loop to display each temperature. Then calculate the highest and lowest temperature in the tuple.

```python
## Exercise 2: Weekend Temperature with a Tuple

temperatures = (24, 31, 27, 23, 33, 28)

print("temperatures are : ")
for temp in temperatures:
    print(temp, "C")

highest_temp = temperatures[0]
lowest_temp = temperatures[0]

for temp in temperatures:
    if temp > highest_temp:
        highest_temp = temp
    if temp < lowest_temp:
        lowest_temp = temp

print("highest temperature is : ", highest_temp)
print("lowest temperature is : ", lowest_temp)
```

### Problem_3  -  Ask the user to enter a word or short sentence. Display the number of characters, the number of vowels (a, e, i, o, u), and the number of spaces. Use a for loop to examine each character.

```python
## Exercise 3: Word Analyzer

text = input("enter a word or short sentence : ")

characters = 0
vowels = 0
spaces = 0

for ch in text:
    characters += 1
    if ch.lower() in "aeiou":
        vowels += 1
    if ch == " ":
        spaces += 1

print("number of characters : ", characters)
print("number of vowels : ", vowels)
print("number of spaces : ", spaces)
```

### Problem_3  -  Write a function called calculate_grade(marks) that receives a list of marks. The function should calculate the average and return a grade using these rules: A for average ≥ 80, B for ≥ 70, C for ≥ 60, D for ≥ 50, and F for below 50. Call the function using a list of 5 marks.  

```python
## Exercise 4: Grade Calculator Function

def calculate_grade(marks):
    total = 0
    for mark in marks:
        total += mark
    average = total / len(marks)

    if average >= 80:
        grade = "A"
    elif average >= 70:
        grade = "B"
    elif average >= 60:
        grade = "C"
    elif average >= 50:
        grade = "D"
    else:
        grade = "F"
    return average, grade


student_marks = [85, 78, 90, 72, 88]
avg, grade = calculate_grade(student_marks)
print("marks are : ", student_marks)
print("average is : ", avg)
print("grade is : ", grade)
```

### Problem_5  - Create a list of dictionaries representing 4 products. Each dictionary should contain product name, price, and quantity. Use a for loop to calculate the total cost of the cart. Then write a function called apply_discount(total) that returns the final price after a 10% discount when the total is greater than or equal to 2,000; otherwise, no discount is applied. 

```python
## Exercise 5: Shopping Cart Processor

products = [
    {"name": "Notebook", "price": 120, "quantity": 3},
    {"name": "Pen", "price": 20, "quantity": 5},
    {"name": "USB Drive", "price": 850, "quantity": 2},
    {"name": "Mouse", "price": 650, "quantity": 1},
]


def apply_discount(total):
    if total >= 2000:
        return total - total * 0.10     # 10% discount
    return total


cart_total = 0
for item in products:
    cart_total += item["price"] * item["quantity"]

print("total cost of cart : ", cart_total)
print("final price after discount : ", apply_discount(cart_total))

```

### Problem_6  -   Write a program using a while loop that asks the user for a number and prints a countdown from that number to zero. 

```python
## Exercise 6: Countdown

number = int(input("enter a number : "))

while number >= 0:
    print(number)
    number -= 1
```
